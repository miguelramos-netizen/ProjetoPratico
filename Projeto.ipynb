```python
# Importação da biblioteca Pandas
import pandas as pd

print("=== 1. OBTENÇÃO E EXPLORAÇÃO DOS DADOS ===")

# Lendo o arquivo CSV local
try:
  df = pd.read_csv(
      'conjunto-dados.csv', sep=';', quotechar='"', encoding='utf-8-sig'
  )
except:
  df = pd.read_csv(
      'conjunto-dados.csv', sep=';', quotechar='"', encoding='latin-1'
  )

# 1. Identificando características básicas
print(f'Dimensões do dataset (linhas, colunas): {df.shape}')
print(f'Quantidade total de registros: {df.shape[0]}')
print(f'Quantidade de atributos (colunas): {df.shape[1]}')

print('\nTipos de dados:')
print(df.dtypes)

print('\nVerificação de valores ausentes por coluna:')
print(df.isnull().sum())


print('\n=== 3. PREPARAÇÃO DOS DADOS ===')
# Tratando dados ausentes: preenchendo os valores nulos da coluna 'Tags' com o texto 'Sem Tag'
df['Tags'] = df['Tags'].fillna('Sem Tag')

# Criando uma nova coluna (transformação): Média de downloads por recurso de cada dataset
df['Downloads_Por_Recurso'] = (
    df['Quantidade Downloads'] / df['Quantidade Recursos']
).round(2)

print('Verificação de nulos após o tratamento:')
print(df.isnull().sum())
print('\nAmostra dos dados após a transformação:')
print(
    df[
        [
            'Organização',
            'Quantidade Downloads',
            'Quantidade Recursos',
            'Downloads_Por_Recurso',
        ]
    ]
    .head(3)
)


print('\n=== 4. ANÁLISE DOS DADOS ===')

# Estatísticas descritivas básicas da Quantidade de Downloads
print('Estatísticas descritivas (Quantidade de Downloads):')
print(df['Quantidade Downloads'].describe().round(2))

# Aplicando um Filtro: Selecionando apenas datasets que possuem mais de 100 downloads
print('\nFiltro: Datasets com mais de 100 downloads')
df_populares = df[df['Quantidade Downloads'] > 100]
print(f'Quantidade de datasets populares (> 100 downloads): {df_populares.shape[0]}')

# Agrupamento (Groupby): Somando a quantidade total de downloads e recursos por Organização
print('\nAgrupamento: Total de Downloads e Recursos por Organização')
resumo_organizacao = (
    df.groupby('Organização')[['Quantidade Downloads', 'Quantidade Recursos']].sum()
    .reset_index()
)
print(resumo_organizacao)
# Soma total dos downloads dividida pela soma total dos recursos
resultado = df['Quantidade Downloads'].sum() / df['Quantidade Recursos'].sum()
print(f'Resultado: {resultado:.2f}')
```
